# __PROJECT__ — схемы

Источник всех картинок проекта. Каждая ячейка рисует одну схему, сохраняет её в `docs/diagrams/<имя>.svg` и показывает здесь же. В документации: `![...](diagrams/<имя>.svg)`.

- Открыть и править: `./tools/diagrams.ps1 lab` → http://127.0.0.1:8888
- Пересобрать все SVG: `./tools/diagrams.ps1 render`

Подробно — [docs/diagrams.md](../../../../docs/diagrams.md). Пример — `projects/led_clock/docs/diagrams/schematics.ipynb`.

```python
from pathlib import Path

from IPython.display import SVG, display
import schemdraw
import schemdraw.elements as elm

schemdraw.use('svg')
schemdraw.config(bgcolor='white', fontsize=12, font='Arial', margin=0.4)

OUT = Path('.')  # ноутбук выполняется из docs/diagrams/


def save(d, name):
    """Сохранить схему в docs/diagrams/<name>.svg и показать в ноутбуке."""
    path = OUT / f'{name}.svg'
    d.save(str(path))
    display(SVG(filename=str(path)))
```

## Подключение к ESP32

WireViz: заменить пример своими модулями и пинами (должно совпадать с `docs/03-hardware.md` и `config.example.h`).

```python
from wireviz.wireviz import parse

WIRING = '''
options:
  fontname: arial
  bgcolor: WH
  bgcolor_node: WH
  bgcolor_connector: WH
  bgcolor_cable: WH
  bgcolor_bundle: WH

connectors:
  ESP32:
    type: DOIT ESP32 DevKit V1
    pins:      [3V3, GND, 21, 22]
    pinlabels: [3V3, GND, D21, D22]
  MODULE:
    type: модуль (пример)
    pinlabels: [VCC, GND, SDA, SCL]

cables:
  W1: {category: bundle, colors: [RD, BK, YE, GN], show_wirenumbers: false}

connections:
  - [ESP32: [3V3, GND, 21, 22], W1: [1-4], MODULE: [1-4]]
'''

parse(WIRING, output_formats=('svg',), output_dir=str(OUT), output_name='wiring')
display(SVG(filename=str(OUT / 'wiring.svg')))
```
